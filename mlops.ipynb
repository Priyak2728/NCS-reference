```python
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array([[1000], [1500], [2000], [2500]])
y = np.array([50, 75, 100, 125])

model = LinearRegression()
model.fit(X, y)

print("Model Trained")
```

```python
import joblib

joblib.dump(model, "house_price_model.pkl")

loaded_model = joblib.load("house_price_model.pkl")

print("Model Saved and Loaded")
```

```python
prediction = loaded_model.predict([[1800]])[0]

actual_price = 120

error = abs(actual_price - prediction)

print("Prediction:", prediction)
print("Actual:", actual_price)
print("Error:", error)

if error > 20:
    print("Retraining Required")
else:
    print("Model Healthy")
```
